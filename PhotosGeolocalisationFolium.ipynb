####  Construire la carte

!!! info Info
La carte que l'on va construire ci-dessous est centrée sur la mairie de Caen avec un zoom de 15. 

Les dimensions de la carte sont indiquées dans les variables `width` (pour la largeur) et `height` (pour la hauteur) ; elles peuvent être changées si vous le souhaitez.
!!!

!!! question A faire 2
Exécuter la cellule pour construire la carte qui sera stockée dans la variable `carte`.
!!!

```python
latitude_centre = 49.1811
longitude_centre = -0.3712
# on donne la taille de la carte en pixels 
carte = folium.Map(location=[latitude_centre, longitude_centre], zoom_start=15, width=800, height=500)
carte # pour afficher la carte dans le notebook
```

#### Ajouter une icône à l'endroit de la photo
!!! info Info
Le fichier image `exemple.jpg` a été prise à un endroit dont les coordonnées GPS sont $(49.175711, -0.372167)$ ; il est possible de le vérifier en examinant les métadonnées de ce fichier.


Dans le code, les variables `latitude` et `longitude` contiennent respectivement la latitude et la longitude de la photo de l'exemple.
!!!

!!! question A faire 3 
Exécuter la cellule pour ajouter à la carte `carte` une icône à l'emplacement où a été prise la carte.
!!!


```python
latitude = 49.175711
longitude = -0.372167
icon = folium.features.CustomIcon(icon_image = "exemple.jpg", icon_size=(30, 30))
folium.Marker([latitude, longitude], tooltip = "exemple", icon=icon).add_to(carte)

carte # pour afficher la carte dans le notebook
```
